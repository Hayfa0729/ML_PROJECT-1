```python
print("Categories in 'gender' variable: " , end = " ")
print(df['gender'].unique())

print("Categories in 'race_ethnicity' variable: " , end = " ")
print(df['race_ethnicity'].unique())

print("Categories in 'parental_level_of_education' variable: " , end = " ")
print(df['parental_level_of_education'].unique())

print("Categories in 'lunch' variable: " , end = " ")
print(df['lunch'].unique())

print("Categories in 'test_preparation_course' variable: " , end = " ")
print(df['test_preparation_course'].unique())

print("Categories in 'math_score' variable: " , end = " ")
print(df['math_score'].unique())

print("Categories in 'reading_score' variable: " , end = " ")
print(df['reading_score'].unique())

print("Categories in 'writing_score' variable: " , end = " ")
print(df['writing_score'].unique())


```

```python
numeric_features = [
    features for features in df.columns 
    if df[features].dtype != 'O'
    ]
categorical_features = [
    features for features in df.columns 
    if df[features].dtype == 'O'
    ]

print("We have {} numerical features : {} ".format(len(numeric_features), numeric_features))
print("We have {} categorical features : {} ".format(len(categorical_features), categorical_features))
```

```python
df['total_score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average_score'] = df['total_score'] / 3
df.head()
```

```python
reading_full = (df['reading_score'] == 100).sum()
writing_full = (df['writing_score'] == 100).sum()
math_full = (df['math_score'] == 100).sum()

print("Number of students who scored full marks in reading: ", reading_full)
print("Number of students who scored full marks in writing: ", writing_full)
print("Number of students who scored full marks in math: ", math_full)
```

```python
reading_score_20 = (df['reading_score'] <= 20).sum()
print("Number of students who scored 20 or less in reading: ", reading_score_20)

writing_score_20 = (df['writing_score'] <= 20).sum()
print("Number of students who scored 20 or less in writing: ", writing_score_20)

math_score_20 = (df['math_score'] <= 20).sum()
print("Number of students who scored 20 or less in math: ", math_score_20)
```

```python
fig, axis = plt.subplots(1, 3, figsize=(15, 5))

axis[0].hist(df['math_score'], bins=20, color='red', alpha=0.7)
axis[0].set_title('Math Scores')
axis[0].set_xlabel('Score')
axis[0].set_ylabel('Frequency')

axis[1].hist(df['reading_score'], bins=20, color='blue', alpha=0.7)
axis[1].set_title('Reading Scores')
axis[1].set_xlabel('Score')
axis[1].set_ylabel('Frequency')

axis[2].hist(df['writing_score'], bins=20, color='green', alpha=0.7)
axis[2].set_title('Writing Scores')
axis[2].set_xlabel('Score')
axis[2].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
