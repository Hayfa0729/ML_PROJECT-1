# Conclusion: Average Score Distribution by Race/Ethnicity

The histogram shows the distribution of students' average scores across different race/ethnicity groups. Group C has a noticeable concentration of scores around 60–75, while Group D shows a concentration around 65–80. Group A appears to have fewer observations overall, with scores distributed mainly around the middle range.
```python
plt.figure(figsize = (18 , 8))
plt.subplot(1 , 4 , 1)
plt.title("MATH SCORES")
sns.violinplot(y = 'math_score' , data = df , color = 'red' , linewidth = 3)
plt.subplot(1 , 4 , 2)
plt.title("READING SCORES")
sns.violinplot(y = 'reading_score' , data = df , color = 'green' , linewidth = 3)
plt.subplot(1 , 4 , 3)
plt.title("WRITING SCORES")
sns.violinplot(y = 'writing_score' , data = df , color = 'blue' , linewidth = 3)
plt.subplot(1 , 4 , 4)
plt.title("AVERAGE SCORES")
sns.violinplot(y = 'average_score' , data = df , color = 'red' , linewidth = 3)
plt.show()
```

Think of a violin plot as a distribution shown sideways.
- Wider section: More scores are concentrated around that value.
- Narrower section: Fewer scores are concentrated around that value.
- Vertical position: Shows the score value.
- Overall height: Shows the range of values in the distribution.
Unlike a box plot, a violin plot uses a smoothed density estimate to show where observations are concentrated.

### Conclusion: Distribution of Student Scores Using Violin Plots

The violin plots show the distributions of math, reading, writing, and average scores among students. Math scores have a median of approximately 66, while reading and writing scores have medians around 70. This suggests that students generally perform somewhat better in reading and writing than in mathematics.

The widest sections of the violins indicate that scores are concentrated mainly in the middle ranges, rather than at the extreme minimum or maximum values. The average-score distribution also shows that most students' overall performance lies around the middle-to-upper score range.

Overall, the plots suggest differences in score distributions across subjects, with math having a slightly lower central score than reading and writing. Further numerical analysis can confirm these observations.

In a violin plot:
- Width tells you where scores are more densely concentrated.
- White line indicates the median in this Seaborn plot.
- Dark inner box represents the interquartile range (the middle 50% of scores).
- Thin lines show the remaining range represented by the inner box plot's whiskers, depending on the plotting settings.
The long, thin tails suggest that some students have scores far from the central concentration. They don't automatically mean those observations are outliers.
